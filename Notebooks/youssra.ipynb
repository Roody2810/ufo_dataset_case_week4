```python
import pandas as pd
import plotly.express as px

# Bestanden inladen
drugs = pd.read_csv("../data/drugs.csv")
ufo = pd.read_csv("../data/ufo_sighting_data.csv", low_memory=False)

# -------------------------
# 1. UFO-meldingen per staat tellen
# -------------------------

ufo_per_state = (
    ufo.groupby("state/province")
    .size()
    .reset_index(name="UFO_count")
)

# -------------------------
# 2. Drugsdata: alleen 18+
# -------------------------

# 18-25 + 26+ = totale groep 18+
drugs["Marijuana_18plus"] = (
    drugs["Rates.Marijuana.Used Past Year.18-25"] +
    drugs["Rates.Marijuana.Used Past Year.26+"]
)

# -------------------------
# 3. Per staat één waarde maken
# -------------------------

drugs_per_state = (
    drugs.groupby("State")["Marijuana_18plus"]
    .mean()
    .reset_index()
)

# -------------------------
# 4. Namen van staten gelijk maken
# -------------------------

# -------------------------
# 4. Staatnamen gelijk maken
# -------------------------

state_codes = {
    "alabama": "al",
    "alaska": "ak",
    "arizona": "az",
    "arkansas": "ar",
    "california": "ca",
    "colorado": "co",
    "connecticut": "ct",
    "delaware": "de",
    "district of columbia": "dc",
    "florida": "fl",
    "georgia": "ga",
    "hawaii": "hi",
    "idaho": "id",
    "illinois": "il",
    "indiana": "in",
    "iowa": "ia",
    "kansas": "ks",
    "kentucky": "ky",
    "louisiana": "la",
    "maine": "me",
    "maryland": "md",
    "massachusetts": "ma",
    "michigan": "mi",
    "minnesota": "mn",
    "mississippi": "ms",
    "missouri": "mo",
    "montana": "mt",
    "nebraska": "ne",
    "nevada": "nv",
    "new hampshire": "nh",
    "new jersey": "nj",
    "new mexico": "nm",
    "new york": "ny",
    "north carolina": "nc",
    "north dakota": "nd",
    "ohio": "oh",
    "oklahoma": "ok",
    "oregon": "or",
    "pennsylvania": "pa",
    "rhode island": "ri",
    "south carolina": "sc",
    "south dakota": "sd",
    "tennessee": "tn",
    "texas": "tx",
    "utah": "ut",
    "vermont": "vt",
    "virginia": "va",
    "washington": "wa",
    "west virginia": "wv",
    "wisconsin": "wi",
    "wyoming": "wy"
}

drugs_per_state["State"] = (
    drugs_per_state["State"]
    .str.lower()
    .map(state_codes)
)

ufo_per_state["state/province"] = (
    ufo_per_state["state/province"]
    .str.lower()
    .str.strip()
)

# -------------------------
# 5. Data combineren
# -------------------------

data = drugs_per_state.merge(
    ufo_per_state,
    left_on="State",
    right_on="state/province",
    how="inner"
)

print("Aantal rijen na merge:", len(data))
print(data.head())

# -------------------------
# 6. Scatterplot maken
# -------------------------
fig = px.scatter(
    data,
    x="Marijuana_18plus",
    y="UFO_count",
    hover_name="State",
    title="Marihuanagebruik (18+) vs. UFO-meldingen per staat",
    labels={
        "Marijuana_18plus": "Marihuanagebruik (%)",
        "UFO_count": "Aantal UFO-meldingen"
    }
)

fig.show()
```

```python
# Datum omzetten naar datetime
ufo["Date_time"] = pd.to_datetime(ufo["Date_time"], errors="coerce")

# Jaar uit de datum halen
ufo["Year"] = ufo["Date_time"].dt.year

# Aantal UFO-meldingen per jaar
ufo_per_year = (
    ufo.groupby("Year")
    .size()
    .reset_index(name="UFO_count")
)

# Sorteren op jaar
ufo_per_year = ufo_per_year.sort_values("Year")

print(ufo_per_year)
```

```python
fig = px.line(
    ufo_per_year,
    x="Year",
    y="UFO_count",
    markers=True,
    title="Aantal UFO-meldingen per jaar",
    labels={
        "Year": "Jaar",
        "UFO_count": "Aantal UFO-meldingen"
    }
)

fig.show()
```
